# rerun the lab's start-of-input correction

This runs one published file, `rerun_ungated.py`, on Qwen2.5-0.5B-Instruct. It checks whether a one-word input sitting alone at position 0 gives a hidden state far larger than the same word in seven other templates, and whether that one state then holds nearly all the variance across 96 words.
The lab's numbers for this model, from sealed test 1 (line P6 of `RESULT_start_token_artifact_2026_11_10.md`): norm ratio 79.9x, top-direction share 0.9885, AUC 1.0000, share with the bare template dropped 0.1038.
The script prints your numbers beside these and says YES or NO for the sealed bars.
No account with the lab and no licence are needed. Run the cells in order.
About 1 GB of download and a few minutes on a processor, as measured on the lab's own machine. The lab has not run this notebook on Colab, so the time here is not known.

```python
# the commit of main that holds the published script: change nothing here, so you run exactly those bytes
COMMIT = "4bbe4b3e27aaa5bc429a058ca7b788ef5b8e9b4f"
EXPECTED_SHA256 = "39cabff71dc98c511c8653ab797be8a4634458b4b6c2dbb22f8696e03853d92c"

import hashlib, platform, urllib.request
import torch, transformers

url = "https://raw.githubusercontent.com/fathom-lab/styxx/" + COMMIT + "/papers/real-convergence/artifact_2026_11_10/rerun_ungated.py"
data = urllib.request.urlopen(url, timeout=30).read()
with open("rerun_ungated.py", "wb") as f:
    f.write(data)
digest = hashlib.sha256(data).hexdigest()
print("commit:      ", COMMIT)
print("sha256:      ", digest)
print("matches the published file:", digest == EXPECTED_SHA256)
print("machine:     ", platform.platform(), platform.processor() or "")
print("torch:       ", torch.__version__, " transformers:", transformers.__version__)

!python rerun_ungated.py --cpu
```
